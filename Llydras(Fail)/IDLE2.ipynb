```python
class A:
    def __init__(self, cls):
        self.cls = cls
    def __getitem__(self, x):
        return self.cls(x)

class Type:
    def __init__(self, _type):
        self._type = _type
    def __repr__(self):
        return f"{type(self).__name__[1:-1]}[{self._type}]"

class _Maybe_(Type):
    def __call__(self, x):
        if x == None:
            return None
        else:
            return self._type(x)
class _List_(Type):
    def __call__(self, x):
        return [self._type(w) for w in list(x)]
        
Maybe = A(_Maybe_)
List = A(_List_)
```

```python
class HigherOrderGate:
    def __init__(self, fn, n):
        self.fn = fn
        self.n = n
    def __getitem__(self, xs):
        if not isinstance(xs, tuple):
            xs = (xs, )
        if len(xs) == self.n:
            return HigherOrderGateSqBra(self.fn, xs)
        else:
            raise Exception(':(')
    def __call__(self, *args, **kwargs):
        return HigherOrderGatesRdBra(self.fn, args, kwargs)
    def __repr__(self):
        return f'HigherOrderGate: {self.fn.__name__}'

class HigherOrderGateSqBra:
    def __init__(self, fn, args):
        self.fn = fn
        self.args = args
    def __call__(self, *args, **kwargs):
        return self.fn(*self.args, *args, **kwargs)
    def __repr__(self):
        return f'HigherOrderGateSqBrac: {self.fn.__name__}'

class HigherOrderGatesRdBra:
    def __init__(self, fn, args, kwargs):
        self.fn = fn
        self.args = args
        self.kwargs = kwargs
    def __getitem__(self, xs):
        if not isinstance(xs, tuple):
            xs = (xs, )
        return self.fn(*xs, *self.args, **self.kwargs)
    def __repr__(self):
        return f'HigherOrderGatesRdBra: {self.fn.__name__}'

class _hog_:
    def __init__(self, num = 1):
        self.num = num
    def __getitem__(self, num):
        return _hog_(num)
    def __call__(self, fn):
        return HigherOrderGate(fn, self.num)

hog = _hog_()
```

```python
# matrix.py

class Matrix:
	def __init__(self, rows: int, columns: int = 1):
		self.rows = rows
		self.columns = columns

		self._raw = [0 for i in range(rows * columns)]
	def add_column(self, columns: int = 1):
		self._raw.extend(0 for i in range(self.rows * columns))
		self.columns += columns
	def __setitem__(self, coords, value):
		y, x = coords
		index = self._get_index(y, x)
		self._raw[index] = value
	def __getitem__(self, coords):
		y, x = coords
		index = self._get_index(y, x)
		return self._raw[index]
	def _get_index(self, y, x):
		if y < 0 or y > self.rows:
			raise IndexError('Y index out of range.')
		if x < 0 or x > self.columns:
			raise IndexError('X index out of range.')

		return self.rows * x + y
	def row(self, y):
		i = self._get_index(y, 0)
		j = self._get_index(y, self.columns)
		return self._raw[i : j : self.rows]
	def column(self, x):
		i = self._get_index(0, x)
		j = self._get_index(self.rows, x)
		return self._raw[i : j]
	def __str__(self):
		cols = []
		L = 0
		for i in range(self.rows):
			cols.append([str(x) for x in self.row(i)])
			l = max(len(x) for x in cols[-1])
			L = max(L, l)

		out = ''
		for col in cols:
			for x in col:
				out += (L - len(x)) * ' ' + x + ' '
			out += '\n'
		return out
```
