# Sedryx

*A functional, compositional quantum circuit library for Python.*

Sedryx is an experimental quantum computing library designed to prioritize **expressive syntax**, **mathematical clarity**, and **modular circuit design**. Built for researchers, students, and enthusiasts alike, Sedryx reimagines circuit construction as a form of algebraic expression — fluent, symbolic, and fully programmable.

Unlike conventional libraries where circuits are passive data structures, Sedryx treats circuits as **first-class citizens**: composable objects that can be chained, transformed, and embedded seamlessly within one another.

## Core Ideas

Sedryx is based on a few simple but powerful principles:

- **Circuits are constructed explicitly.** A call like `Circuit(2)` returns a new circuit over two quantum bits; `Circuit(-1)` over one classical bit. Sign and type encode semantic intent. For readability, string labels like `'q:x'` or `':c'` are also supported.
  
- **Bits are first-class and unpackable.** `x, y = circuit.all_bits()` gives you named handles for each bit in the system, which you can manipulate and compose.

- **Gates are applied using `>>`, and bit groups are composed with `&`.** These two operators, inspired by functional composition and tensor product, allow both imperative and point-free circuit definitions:
  
  ```python
  x >> H >> X            # applies gates to bit x
  (x & y) >> Cnot        # entangled application
  bell = (H & I) >> Cnot # point-free definition of a Bell pair
Circuits are composable. Entire circuits can be applied like gates to other circuits. Their structure is embedded unless abstracted using .gate(name), which hides internal bits and ancillae.
Immutability by default. Circuit transformations return new objects. Only direct bit operations (>>) mutate the parent circuit, making composition safe and predictable.
High-level abstractions via decorators. Functions can be promoted to general gates using @arity or @signature, enabling definitions like:
@arity
def QFT(n):
    ...
    return circuit.gate(f"QFT({n})")
Which can be applied without manual arity specification: (a & b & c) >> QFT.
Why Sedryx?

Sedryx is a tool for those who think of quantum computation structurally — not as a sequence of low-level instructions, but as a mathematical object to be reasoned about, composed, and transformed. It is designed to scale from toy problems to symbolic circuit definitions suitable for embedding in papers, notebooks, or larger frameworks.

Status

Sedryx is currently in early development and experimental use. The API is subject to refinement, guided by clarity, compositional elegance, and the practical needs of researchers.

If you're interested in quantum computation as a language of structure, you may find Sedryx a compelling companion.

````python
import re
from datetime import datetime

def parse_whatsapp_chat(filepath):
    pattern = re.compile(r'^\[(\d{2}/\d{2}/\d{4} \d{2}:\d{2}:\d{2})\] (.*?): (.*)')
    messages = []

    with open(filepath, 'r') as file:
        current_msg = None

        for line in file:
            line = line.strip()
            match = pattern.match(line)

            if match:
                if current_msg:
                    messages.append(current_msg)

                timestamp_str, sender, message = match.groups()
                timestamp = datetime.strptime(timestamp_str, '%d/%m/%Y %H:%M:%S')

                current_msg = {
                    'timestamp': timestamp,
                    'sender': sender,
                    'message': message
                }
            else:
                # Continuation of the previous message
                if current_msg:
                    current_msg['message'] += '\n' + line

        if current_msg:
            messages.append(current_msg)

    return messages
````

### Storage Backend
Although the system is designed to be agnostic to the choice of storage backend, a relational database such as SQLite is particularly well-suited due to its simplicity and efficiency. Below is the proposed schema used to persist folder metadata, file data, and lookup tables. All cryptographic operations are performed externally; the database only stores encrypted or hashed data.

#### Database Schema
##### `folder` Table
| Column          | Type    | Description                                                             |
| --------------- | ------- | ----------------------------------------------------------------------- |
| `id`            | INTEGER | Unique identifier for the folder                                        |
| `name`          | BLOB    | Encrypted folder name using the shared key $S$.                         |
| `password_salt` | BLOB    | Encrypted password salt using $S$.                                      |
| `file_salt`     | BLOB    | Encrypted salt for file lookups using $S$.                              |
| `child_salt`    | BLOB    | Encrypted salt for child lookups using $S$.                             |
| `key_salt`      | BLOB    | Public salt used in key lookups (unencrypted)                           |

| `key2`          | BLOB    | Public key for the complementary mode (duress or primary) (unencrypted) |

````python
alphabet = 'abcdefghijklmnopqrstuvwxyzABCDEFGHIJKLMNOPQRSTUVWXYZ0123456789 .,!()[]{}?:;'
key = {x: i for i, x in enumerate(alphabet)}
k = [x for x in alphabet]
matrix = [shift(k, -i) for i in range(len(alphabet))]
def get(x, y):
    return matrix[key[x]][key[y]]
````

````python
def encrypt(secret, message):
    q = len(secret)
    cipher = ''
    for i, x in enumerate(message):
        cipher += get(secret[i % q], x)
    return cipher

def decrypt(secret, cipher):
    q = len(secret)
    plain = ''
    for i, x in enumerate(cipher):
        z = matrix[key[secret[i % q]]].index(x)
        plain += alphabet[z]
    return plain
````

````python
def repeat(xs, l):
    q, r = divmod(l, len(xs))
    return xs * q + xs[:r]
    
def xor(x, y):
    sx, sy = str(x), str(y)
    sy = repeat(sy, len(sx))
    cipher = ''
    for i in range(len(sx)):
        cipher += str((int(sx[i]) + int(sy[i])) % 10)
    return cipher

def ixor(x, y):
    sx, sy = str(x), str(y)
    sy = repeat(sy, len(sx))
    cipher = ''
    for i in range(len(sx)):
        cipher += str((int(sx[i]) - int(sy[i])) % 10)
    return cipher
````

````python
def Round(key, fn, left, right):
    return right, xor(left, fn(key, right))

def unRound(key, fn, left, right):
    return right, ixor(left, fn(key, right))
````

````python
def F(key, block):
    #print('Block: ', block)
    L = len(block)
    block = xor(xor(block, shift(block, 2)), shift(block, -2))
    K = xor(xor(key, shift(key, 2)), shift(key, -2))
    #print('Xor')
    #print('Block: ', block)
    #print('Key: ', K)
    M = interlace(K, shift(K, 2))
    m = xor(M[:len(key)], M[len(key):])
    key = repeat(m, L)
    cipher = xor(block, key)
    return cipher #xor(cipher[L:], shift(cipher[:L], 2))
````

````python
lookup = {chr(97 + i) : str(10 + i) for i in range(26)}
lookup.update({x : str(36 + i) for i, x in enumerate(('.', ',', '!', '?', '\''))})
lookup.update({str(i) : str(40 + i) for i in range(1, 10)})
lookup.update({'0' : '50', ' ' : '51'})
lookup_str = str.maketrans(lookup)

de = {}
for i, j in lookup.items():
    de[j.zfill(2)] = i
````

````python
def encode(message):
    message = message.lower()
    #out = [lookup[x] for x in message]
    return message.translate(lookup_str)

def decode(message):
    #out = [lookup[x] for x in message]
    return ''.join(de.get(x, ' ') for x in wrap(message, 2))
````

````python
def Encrypt(e, keys):
    l = len(e) // 2
    left, right = e[:l], e[l:]
    for key in keys:
        left, right = Round(key, F, left, right)
    return right, left

def Decrypt(e, keys):
    l = len(e) // 2
    K = len(keys)
    left, right = e[:l], e[l:]
    for i in range(1, K + 1):
        key = keys[K - i]
        left, right = unRound(key, F, left, right)
    return right, left

def encrypt(message, key):
    keys = encode(key).split('51')
    return ''.join(Encrypt(encode(message), keys))

def decrypt(message, key):
    keys = encode(key).split('51')
    return decode(''.join(Decrypt(message, keys)))
````
